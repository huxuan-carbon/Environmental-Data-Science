```python
passengers=[320,450,510,620,580]
for p in passengers:
    print(p)
```

```python
for p in passengers:
    if p>=500:
        print(f"找到超标客流：{p}")
        break
```

```python
total=0
i=0
while passengers[i]<=600 and i<=len(passengers):
    total=total+passengers[i]
    i=i+1
print(f"限流前累计客流：{total}")
```
