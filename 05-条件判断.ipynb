```python
ridership=230000
is_transfer_station=True
if ridership>200000 and is_transfer_station:
    print("重点保障站：客流量大且是换乘站")
elif ridership>200000 or is_transfer_station:
    print("需要关注的站")
else:
    print("非换乘普通站")
```

```python
if not is_transfer_station:
    print("非换乘站")
else:
    print("换乘站")
```

```python
daily_ridership={"西直门":250000,"动物园":180000,"国家图书馆":220000,"魏公村":150000,"人民大学":210000}
```

```python
electricity_kwh=3200
grid_factor=0.581
carbon_emission=electricity_kwh*grid_factor
print(f"该站日碳排放量:{carbon_emission:.1f} kg CO2")
```

```python
if carbon_emission<1000:
    print("低碳排放站")
elif carbon_emission<2000:
    print("中等碳排放站")
else:
    print("高碳排放站，建议排查节能空间")
```
