```python
def rect_area(length,width):
    return length*width
l=float(input("请输入长（米）："))
w=float(input("请输入宽（米）："))
area=rect_area(l,w)
price_per_sqm=80
total_cost=area*price_per_sqm
print(f"面积是：{area}平方米,装修费预估：{total_cost}元")
```

```python
def calc_carbon(electricity,factor=0.581):
    return electricity*factor
print(calc_carbon(3200))
```

```python
def total_flow(*flows):
    total=0
    for f in flows:
        total+=f
    return total
data=input()
nums=[float(x) for x in data.split(',')]
print(total_flow(*nums))
```
